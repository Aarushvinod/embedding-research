# UCDEmbed — STS-focused study (UCD-vector students)

The UCD counterpart to `byteembed_sts_a100.ipynb`: tests whether **UCD property-vector** embeddings work for STS *when trained for it*. Supervised STS fine-tuning with the **CoSENT** rank objective.

**8 languages, all with real native graded `sentence1/sentence2/score` train data:**
- **6 low-resource:** Amharic, Hausa, Kinyarwanda, Telugu, Marathi, Algerian Arabic (SemRel24STS)
- **2 anchors:** English (SemRel24STS), Chinese (C-MTEB/STSB)

Train on `train`+`dev`, **evaluate on `test` only**. Each UCD student is **warm-started from its SONAR-distilled checkpoint** (`checkpoints/ucd-*_mean.pt`) — important here because that transfers the **trained UCD featurizer** instead of starting it cold (its cold-start handicap is the main reason UCD trailed byte in the distillation runs). Identical recipe / pooling to the byte notebook, so byte-vs-UCD STS is a clean comparison. Reports per-language Spearman, raw and with **whitening**. Smoke cell first.

### 5. Full STS run — UCD small / base / large
